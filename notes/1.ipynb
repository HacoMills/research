```python
import numpy as np

# 示例数据：策略收益率（月收益率）
returns = np.array([0.02, 0.03, -0.01, 0.04, 0.01, 0.02, -0.02, 0.03, 0.01, 0.02, 0.03, 0.01])

# 无风险收益率（年化）
risk_free_rate = 0.03  # 假设无风险收益率为3%

# 计算年化收益率
annualized_return = np.prod(1 + returns) ** (12 / len(returns)) - 1

# 计算年化标准差
annualized_std = np.std(returns) * np.sqrt(12)

# 计算夏普比率
sharpe_ratio = (annualized_return - risk_free_rate) / annualized_std

print(f"年化收益率: {annualized_return:.4f}")
print(f"年化标准差: {annualized_std:.4f}")
print(f"无风险收益率: {risk_free_rate:.4f}")
print(f"夏普比率: {sharpe_ratio:.4f}")
```
